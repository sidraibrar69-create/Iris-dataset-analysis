**Import Required Libraries**


```python
import pandas as pd  #pandas used for working with dataset handling, loading and analyzing
import seaborn as sns #seaborn used for beaitiful visualizaton and built-in datastets
import matplotlib.pyplot as plt  #used for plotting graph, base plotting library

```

**Load the dataset**

```python
iris=sns.load_dataset('iris')
```

# View Dataset Shape

```python
print(iris.shape)
```

```python
print(iris.columns)
```

```python
print(iris.head())
```

```python
print(iris.info())
```

```python
print(iris.describe())
```

```python
sns.scatterplot(
    x="sepal_length",
    y="sepal_width",
    hue="species",
    data=iris
)

plt.title("Sepal Length vs Sepal Width")
plt.show()
```

```python
iris.hist(figsize=(10, 8))
plt.show()
```

```python
plt.figure(figsize=(8, 6))

sns.boxplot(data=iris)

plt.title("Box Plot of Iris Features")
plt.show()
```
